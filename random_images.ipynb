```python
import matplotlib.pyplot as plt
import numpy as np
img = np.random.randint(0, 256, (20, 20, 3), dtype=np.uint8)
plt.imshow(img)
plt.axis('off')
plt.title('Random Image')
plt.show()
```

```python
img = np.random.randint(0, 255, (20, 20), dtype=np.uint8)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title('Random Image - Grayscale')
plt.show()
```

```python
import random
img = np.full((20, 20), fill_value=255, dtype=np.uint8)
for i in range(20):
    for j in range(20):
       img[i, j] = 0 if random.randint(0, 1) else 220
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.show()
```

```python
from IPython.display import clear_output, display
import time

img = np.full((20, 20), fill_value=255, dtype=np.uint8)
for frame in range(50):
    for i in range(20):
        for j in range(20):
            img[i, j] = 0 if random.randint(0, 1) else 220
    plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    plt.axis('off')
    display(plt.gcf())
    clear_output(wait=True)
    time.sleep(0.1)
```

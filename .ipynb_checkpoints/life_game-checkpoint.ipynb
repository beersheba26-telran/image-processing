```python
import numpy as np
import matplotlib.pyplot as plt
from life_game_matrix import getInitialMatrix, getNextMatrix
SHAPE_0 = 20
SHAPE_1 = 20
LIMIT = 100
img: np = getInitialMatrix(SHAPE_0, SHAPE_1)
img_next: np = None
iteration: int = 0
plt.ion()
fig, ax = plt.subplots()
plot = ax.imshow(img, cmap="gray", vmin=0, vmax=1)

while not np.array_equal(img, img_next := getNextMatrix(img)) and iteration < LIMIT:
    img = img_next
    plot.set_array(img)
    fig.canvas.draw()
    fig.canvas.flush_events()
    iteration += 1
plt.ioff()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
width = 100
height = 100
img = np.zeros((width,height,3), np.uint8)
cv2.line(img,(0,0),(width-1,height-1),(255,0,0),1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('line - diagonally')
plt.show()
```

```python
img = np.zeros((width,height,3), np.uint8)
cv2.circle(img,(width//2,height//2),30,(0,255,0),-1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('circle - center')
plt.show()
```
